## Modelling Preprocessing

Preprocess the dataset after splitting into training and testing sets
to avoid data leakage during imputation.

```python
# Create a copy of the original dataset
model_data = data.copy()

# Convert suspicious zeros into missing values
model_data[zero_as_missing] = model_data[zero_as_missing].replace(0, pd.NA)

# Separate features and target
X_model = model_data.drop('Outcome', axis=1)
y_model = model_data['Outcome']
```

```python
X_train_model, X_test_model, y_train_model, y_test_model = train_test_split(
    X_model,
    y_model,
    test_size=0.20,
    random_state=42,
    stratify=y_model
)
```

```python
training_medians = X_train_model[zero_as_missing].median()

training_medians
```

```python
# Fill missing values in the training data
for column in zero_as_missing:
    X_train_model[column] = X_train_model[column].fillna(
        training_medians[column]
    )

# Fill missing values in the testing data
# Use the same medians calculated from training data
for column in zero_as_missing:
    X_test_model[column] = X_test_model[column].fillna(
        training_medians[column]
    )
```

```python
print("Training missing values:")
print(X_train_model.isnull().sum())

print("\nTesting missing values:")
print(X_test_model.isnull().sum())
```
